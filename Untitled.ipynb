```python
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

fig, ax = plt.subplots()

def update(i):
    ax.clear()
    ax.plot([0, i], [0, i])
    ax.set_title(f"frame {i}")
    return ()

anim = FuncAnimation(fig, update, frames=10, interval=300, blit=False)

HTML(anim.to_jshtml())
```
