```python
from hexgrid import HexCoord, HexGrid, build_obstacle_map, VelocityState
from hexstar import HStarProblem, HStarSearch, h_cost_travel_time, construct_full_solution, BidHStarSearch
from hexplot import plot_hex_grid_2
```

#### Map parameters

```python
center = HexCoord(0,0)
start = HexCoord(5,0)
goal = HexCoord(0,-5)
radius = 8
hex_size = 1
seed = 0
num_obs = 10
```

#### Agent parameters

```python
a_max = 3 # acceleration ability
a_min = 1 # deceleration ability
a_lat = 7 # maximum lateral g-force allowed by the agent
```

#### Build obstacle set

```python
obstacles = set(
    build_obstacle_map(
        center=center,
        radius=radius,
        seed=seed,
        exclude=[start, goal],
        types=[
            ("random_n", {"n": 10})
        ],
    ).keys()
)
print(obstacles)
```

#### Initialize problem class

```python
hex_grid = HexGrid(hex_size, obstacles)

start_v = VelocityState(0, None) # 0 speed, unspecified direction (allows expansion in any direction at root)
goal_v = VelocityState(0, None) # matches any direction as long as location matches. 

problem = HStarProblem(
    hex_grid,
    start,
    goal,
    a_max,  
    a_min,  
    start_v = start_v,
    goal_v = goal_v,
    collision_radius = 0,
    heuristic = h_cost_travel_time,
    
)
```

```python
plot_hex_grid_2(
    obstacles,
    start,
    center,
    radius,
    goal=goal,
    figsize=(25, 25),
    show_coords=False,
    linewidth=1,
    draw_axes=True,
    show_axis_legend=True,
)
```

#### Regular H* search example

```python
hs = BidHStarSearch(
    problem = problem,
    bidirectional = False,
)
fsolution = hs.search()
f_soln_path = construct_full_solution(fsolution)
```

```python
plot_hex_grid_2(
    obstacles,
    start,
    center,
    radius,
    goal=goal,
    figsize=(5, 5),
    show_coords=False,
    linewidth=1,
    path = f_soln_path,
    draw_axes=True,
    show_axis_legend=True,
)
```

#### Bidirectional H* search example

```python
bihs = BidHStarSearch(
    problem = problem
)
bidsolution = hs.search()
bid_soln_path = construct_full_solution(bidsolution)
```
