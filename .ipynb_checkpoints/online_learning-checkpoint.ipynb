```python
from hexgrid import HexCoord, HexGrid, build_obstacle_map, VelocityState, get_neighbors_at_radius, hdist
from hexstarrl import ReplayBuffer, ExperienceCollector, EnvironmentStore, HUNetTDTrainer, HUNetStateEncoder, ValueNetwork_HUNet, RLState

from hexstar import HStarProblem, HStarSearch, h_cost_travel_time_euclidean, h_cost_travel_time_manhattan, construct_full_solution, BidHStarSearch, calculate_path_smoothness, travel_time_euclidean
from hexplot import plot_hex_grid_2

from tqdm.auto import tqdm

from dataclasses import dataclass

import random
import torch
from torch.utils.data import Dataset, DataLoader
import matplotlib.pyplot as plt

from collections import defaultdict
import torch.nn as nn
```

```python
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)
print("Device:", device)
```

```python
# ============================================================
# H* PROBLEM / SEARCH DISTRIBUTION
# ============================================================

NUM_RUNS = 1
FIRST_RUN = 0

CENTER = HexCoord(0, 0)
SAFE_RADIUS = 5
HEX_SIZE = 1


map_type_config = {
    "random_n": {
        "obs_density_min": 0.10,
        "obs_density_max": 0.40,
    },

    "clustered_n": {
        "obs_density_min": 0.10,
        "obs_density_max": 0.40,
        "min_cluster_size": 15,
        "max_cluster_size": 100,
        "exclude_radius": SAFE_RADIUS,
    },
}


search_param_grid = {
    "map_type": [
        "random_n",
        "clustered_n",
    ],

    "search_type": [
        "unidirectional",
    ],

    "heuristic_name": [
        "manhattan",
        "euclidean",
    ],

    "radius": [
        15,
        25,
        35,
    ],

    "ay_window_ms": [
        3,
        7,
    ],

    "jps_horizon": [
        1,
        3,
        5,
        10,
        None,
    ],

    "n_step": [
        1,
        2,
        5,
    ],

    "a_min": [2],
    "a_max": [1],

    "join_tolerance": [50],
    "state_key_mode": ["location"],
}
```

```python
# ============================================================
# RL MODEL PARAMETERS
# ============================================================

spatial_channels = 16
state_features = 5


# ============================================================
# REPLAY BUFFER PARAMETERS
# ============================================================

replay_capacity = 2_000_000
solution_fraction = 0.20


# ============================================================
# RL TRAINING PARAMETERS
# ============================================================

learning_rate = 1e-3

gamma = 1.0

target_update_every = 100

solution_weight = 0.80
nonterminal_weight = 0.20

batch_size = 16

training_steps_per_search = 5
```

```python
def random_start_goal(center, radius, min_distance_ratio=0.5):
    while True:
        q1 = random.randint(-radius + 1, radius - 1)
        r1 = random.randint(
            max(-radius + 1, -q1 - radius + 1),
            min(radius - 1, -q1 + radius - 1)
        )

        q2 = random.randint(-radius + 1, radius - 1)
        r2 = random.randint(
            max(-radius + 1, -q2 - radius + 1),
            min(radius - 1, -q2 + radius - 1)
        )

        start = HexCoord(center.q + q1, center.r + r1)
        goal = HexCoord(center.q + q2, center.r + r2)

        distance = (
            abs(start.q - goal.q)
            + abs(start.r - goal.r)
            + abs((start.q + start.r) - (goal.q + goal.r))
        ) // 2

        if distance >= radius * min_distance_ratio:
            return start, goal
```

```python
# ============================================================
# INITIALIZE RL OBJECTS
# ============================================================

environment_store = EnvironmentStore()

replay_buffer = ReplayBuffer(
    capacity=replay_capacity,
    solution_fraction=solution_fraction,
)

experience_collector = ExperienceCollector(
    replay_buffer=replay_buffer,
    environment_store=environment_store,
)

encoder = HUNetStateEncoder()

value_network = ValueNetwork_HUNet(
    encoder=encoder,
    spatial_channels=spatial_channels,
    state_features=state_features,
)

trainer = HUNetTDTrainer(
    model=value_network,
    replay_buffer=replay_buffer,
    environment_store=environment_store,
    learning_rate=learning_rate,
    gamma=gamma,
    target_update_every=target_update_every,
    solution_weight=solution_weight,
    nonterminal_weight=nonterminal_weight,
)


# ============================================================
# INITIAL STATE
# ============================================================

print("Environment count:", len(environment_store))
print("Replay buffer size:", len(replay_buffer))
print("Grounded samples:", replay_buffer.solution_size)
print("Nonterminal samples:", replay_buffer.nonterminal_size)
print("Trainer steps:", trainer.training_steps)
```

```python
# ============================================================
# BUILD ONE H* PROBLEM
# ============================================================

seed = 0
environment_id = seed

# ------------------------------------------------------------
# Select problem parameters
# ------------------------------------------------------------

map_type = "clustered_n"

radius = 35

heuristic_name = "euclidean"

ay_window_ms = 7

jps_horizon = 5

n_step = 2

a_min = 2
a_max = 1

collision_radius = 0

# Start / goal velocity constraints
start_v = VelocityState(
    magnitude=0,
    direction=None,
)

goal_v = VelocityState(
    magnitude=0,
    direction=None,
)


# ------------------------------------------------------------
# Select heuristic
# ------------------------------------------------------------

heuristics = {
    "manhattan": h_cost_travel_time_manhattan,
    "euclidean": h_cost_travel_time_euclidean,
}

heuristic = heuristics[
    heuristic_name
]


# ------------------------------------------------------------
# Map configuration
# ------------------------------------------------------------

map_config = map_type_config[
    map_type
]

rng = random.Random(
    seed
)

obs_density = rng.uniform(
    map_config["obs_density_min"],
    map_config["obs_density_max"],
)

n_obstacles = int(
    obs_density
    * 3
    * radius
    * (radius + 1)
)


# ------------------------------------------------------------
# Generate start / goal
# ------------------------------------------------------------

start, goal = random_start_goal(
    CENTER,
    radius,
    min_distance_ratio=0.60,
)


# ------------------------------------------------------------
# Build obstacle specification
# ------------------------------------------------------------

if map_type == "random_n":

    obstacle_spec = (
        "random_n",
        {
            "n": n_obstacles,
        },
    )

elif map_type == "clustered_n":

    obstacle_spec = (
        "clustered_n",
        {
            "n": n_obstacles,
            "min_cluster_size": (
                map_config["min_cluster_size"]
            ),
            "max_cluster_size": (
                map_config["max_cluster_size"]
            ),
            "exclude_radius": (
                map_config["exclude_radius"]
            ),
        },
    )

else:
    raise ValueError(
        f"Unsupported map_type: {map_type}"
    )


# ------------------------------------------------------------
# Generate obstacles
# ------------------------------------------------------------

obstacles = set(
    build_obstacle_map(
        center=CENTER,
        radius=radius,
        seed=seed,
        exclude=[
            start,
            goal,
        ],
        types=[
            obstacle_spec,
        ],
    ).keys()
)


# ------------------------------------------------------------
# Add blocked outer boundary
# ------------------------------------------------------------

obstacles.update(
    get_neighbors_at_radius(
        CENTER,
        radius,
    )
)


# ------------------------------------------------------------
# Build HexGrid
# ------------------------------------------------------------

hex_grid = HexGrid(
    hex_size=HEX_SIZE,
    obstacles=obstacles,
    center=CENTER,
    radius=radius,
)


# ------------------------------------------------------------
# Register environment for RL
# ------------------------------------------------------------

environment_store.add(
    environment_id,
    hex_grid,
)


# ------------------------------------------------------------
# Build HStarProblem
# ------------------------------------------------------------

problem = HStarProblem(
    grid=hex_grid,
    start=start,
    goal=goal,

    a_max=a_max,
    a_min=a_min,

    heuristic=heuristic,

    start_v=start_v,
    goal_v=goal_v,

    ay_window_ms=ay_window_ms,
    collision_radius=collision_radius,

    jps_horizon=jps_horizon,
    n_step=n_step,
)


# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

print("Environment ID:", environment_id)
print("Map type:", map_type)
print("Radius:", radius)
print("Start:", start)
print("Goal:", goal)
print("Obstacle density:", obs_density)
print("Requested obstacles:", n_obstacles)
print("Actual obstacles:", len(obstacles))
print("Heuristic:", heuristic_name)
print("n_step:", n_step)
print("JPS horizon:", jps_horizon)
```

```python
plot_hex_grid_2(
    obstacles=obstacles,
    start=start,
    center=CENTER,
    radius=radius + 5,
    hex_size=HEX_SIZE,
    goal=goal,
    figsize=(5, 5),
    show_coords=False,
    linewidth=0,
    #path=solution_path,
    draw_axes=False,
    show_axis_legend=True,
    #save_path=str(save_path),
)
```

```python
# ============================================================
# BUILD H* SEARCH
# ============================================================

hs = BidHStarSearch(
    problem=problem,

    # Search configuration
    bidirectional=False,

    # Experience collection
    experience_collector=experience_collector,
    environment_id=environment_id,

    # Persistent RL training
    online_trainer=trainer,
    online_batch_size=batch_size,
    online_training_steps=training_steps_per_search,
)

for i in range(5):
    
    fsolution = hs.search()
    f_soln_path = construct_full_solution(fsolution)
    print(i, end='|')
```

```python
plot_hex_grid_2(
    obstacles=obstacles,
    start=start,
    center=CENTER,
    radius=radius + 5,
    hex_size=HEX_SIZE,
    goal=goal,
    figsize=(5, 5),
    show_coords=False,
    linewidth=0,
    path=f_soln_path,
    draw_axes=False,
    show_axis_legend=True,
    #save_path=str(save_path),
)
```

```python
import matplotlib.pyplot as plt


def plot_online_training(benchmarks):
    """Plot online value-network training metrics."""

    history = benchmarks.get(
        "online_training_history",
        [],
    )

    if not history:
        print("No online training history.")
        return

    steps = range(
        1,
        len(history) + 1,
    )

    loss = [
        m["loss"]
        for m in history
    ]

    solution_loss = [
        m["grounded_loss"]
        for m in history
    ]

    nonterminal_loss = [
        m["nonterminal_loss"]
        for m in history
    ]

    solution_samples = [
        m["grounded_samples"]
        for m in history
    ]

    nonterminal_samples = [
        m["nonterminal_samples"]
        for m in history
    ]

    fig, axes = plt.subplots(
        1,
        2,
        figsize=(12, 4),
    )

    # ---------------------------------------------------------
    # Loss
    # ---------------------------------------------------------

    axes[0].plot(
        steps,
        loss,
        label="Overall Loss",
    )

    axes[0].plot(
        steps,
        solution_loss,
        label="Solution Loss",
    )

    axes[0].plot(
        steps,
        nonterminal_loss,
        label="Nonterminal Loss",
    )

    axes[0].set_xlabel(
        "Training Step"
    )

    axes[0].set_ylabel(
        "MSE Loss"
    )

    axes[0].set_title(
        "Online Training Loss"
    )

    axes[0].legend()

    # ---------------------------------------------------------
    # Samples contributing to training
    # ---------------------------------------------------------

    axes[1].plot(
        steps,
        solution_samples,
        label="Solution Samples",
    )

    axes[1].plot(
        steps,
        nonterminal_samples,
        label="Nonterminal Samples",
    )

    axes[1].set_xlabel(
        "Training Step"
    )

    axes[1].set_ylabel(
        "Samples"
    )

    axes[1].set_title(
        "Training Sample Composition"
    )

    axes[1].legend()

    plt.tight_layout()
    plt.show()
```

```python
benchmarks = hs.get_benchmarks()

plot_online_training(
    benchmarks
)
```
